```python
t.Age=t.Age.fillna(t.Age.mean())
t.Fare=t.Fare.fillna(t.Fare.mean())
t.Cabin = t.Cabin.fillna("Unknown")
```

```python
## Missing Values Handling

#The Titanic dataset contains missing values in three columns: `Age`, `Fare`, and `Cabin`.

#- **Age:** Missing values were filled using the **mean** because Age is a numerical variable and the median is less affected by extreme values.
#- **Fare:** Missing values were filled using the **mean** because Fare is a numerical variable and may contain extreme values, making median imputation more suitable than mean imputation.
#- **Cabin:** Missing values were replaced with **"Unknown"** because a large number of cabin values are unavailable. Removing these rows would cause unnecessary data loss.

#After applying these methods, the dataset was checked again to confirm that the missing values were handled successfully.
```

```python
t.Sex= t.Sex.astype(str).str.strip().str.lower()

t.Sex = t.Sex.replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
})
```

```python
t.Survived = pd.to_numeric(t.Survived, errors="coerce")
t.Pclass = pd.to_numeric(t.Pclass, errors="coerce")
t.Age = pd.to_numeric(t.Age, errors="coerce")
t.SibSp = pd.to_numeric(t.SibSp, errors="coerce")
t.Parch = pd.to_numeric(t.Parch, errors="coerce")
t.Fare = pd.to_numeric(t.Fare, errors="coerce")
```

```python
## Data Type Correction

#The data types of the Titanic dataset were reviewed and corrected according to the purpose of each column.

#`PassengerId` was converted to a string because it is an identifier rather than a numerical measurement. Text-based columns such as `Name`, `Sex`, `Ticket`, `Cabin`, and `Embarked` were also stored as strings.

#The columns `Survived`, `Pclass`, `Age`, `SibSp`, `Parch`, and `Fare` were converted to appropriate numeric data types to support accurate calculations and analysis.
```

```python
## Value Range Anomaly Detection

#The numerical columns were checked for logically invalid or unusual values.

#The `Age` column was checked to ensure that values were not negative or unrealistically high. The `Fare` column was checked for negative values. The `Pclass` column was expected to contain only values 1, 2, or 3, while the `Survived` column was expected to contain only 0 or 1.

#These checks help identify potential data-entry errors and ensure that the numerical values fall within reasonable ranges.
```

```python
Q1 = t.Age.quantile(0.25)
Q3 = t.Age.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

age_outliers = t[
    (t.Age < lower_bound) |
    (t.Age > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Age Outliers:", len(age_outliers))
```

```python
Q1 = t.Fare.quantile(0.25)
Q3 = t.Fare.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = t[
    (t.Fare < lower_bound) |
    (t.Fare > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", len(fare_outliers))
```

```python
## Outlier Detection Using IQR

#The Interquartile Range (IQR) method was used to identify potential outliers in the `Age` and `Fare` columns.

#The IQR was calculated as the difference between the third quartile (Q3) and the first quartile (Q1). Values below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR were identified as potential outliers.

#The identified observations were inspected before deciding whether to remove, cap, or retain them. This prevents valid extreme observations from being unnecessarily removed.
```

```python
## Outlier Treatment Decision

#The IQR method identified 36 potential outliers in the `Age` column and 55 potential outliers in the `Fare` column.

#These outliers were retained because the extreme values are plausible and may represent genuine observations. For example, higher fares can be associated with first-class passengers or group bookings, while unusual ages can represent actual passengers.

#Since there was no clear evidence that these values were data-entry errors, the outliers were not removed or capped.
```
